```python
import os
import pandas as pd
from dotenv import load_dotenv
from sqlalchemy import create_engine, text
from sentence_transformers import SentenceTransformer

# 1. Load Credentials & Connect
load_dotenv()
DB_USER = os.getenv("DB_USER", "postgres")
DB_PASSWORD = os.getenv("DB_PASSWORD")
DB_HOST = os.getenv("DB_HOST", "localhost")
DB_PORT = os.getenv("DB_PORT", "5432")
DB_NAME = os.getenv("DB_NAME", "quant_vault_db")

engine = create_engine(f"postgresql://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}")

# 2. Encode Vector Search String
search_text = "AI cloud infrastructure growth and generative intelligence"
model = SentenceTransformer("all-MiniLM-L6-v2")
query_embedding = model.encode(search_text).tolist()

# 3. Hybrid SQL Query (pgvector + Fundamentals)
# 3. Hybrid SQL Query (pgvector + Fundamentals - Deduplicated)
hybrid_query = text("""
    SELECT DISTINCT ON (h.company_bk)
        h.company_bk AS "Ticker",
        e.doc_type AS "Doc Type",
        LEFT(e.summary_text, 55) || '...' AS "Press Release Snippet",
        ROUND(CAST(r.priceearningsratio AS numeric), 2) AS "P/E",
        ROUND(CAST(r.returnonequity AS numeric), 4) AS "ROE",
        ROUND(CAST(1 - (e.embedding <=> CAST(:query_vec AS vector)) AS numeric), 4) AS "Similarity Score"
    FROM raw_vault.sat_company_text_embeddings e
    JOIN raw_vault.hub_company h 
        ON e.hk_company_pk = h.hk_company_pk
    JOIN raw_vault.sat_company_financial_ratios r 
        ON h.hk_company_pk = r.hk_company_pk
    ORDER BY h.company_bk, e.embedding <=> CAST(:query_vec AS vector) ASC
    LIMIT 8;
""")

# 4. Render Styled DataFrame Output
df = pd.read_sql(hybrid_query, engine, params={"query_vec": str(query_embedding)})

# Print clean, unstyled DataFrame table
print(f"🔍 Vector Query: '{search_text}'\n")
display(df)
```
